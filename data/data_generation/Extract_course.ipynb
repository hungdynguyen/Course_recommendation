```python
import os
import torch

# ==============================================================================
# 🛠️ BƯỚC 1: CẤU HÌNH MÔI TRƯỜNG
# ==============================================================================
torch_lib_path = os.path.join(os.path.dirname(torch.__file__), 'lib')
os.environ['LD_LIBRARY_PATH']             = torch_lib_path + ':' + os.environ.get('LD_LIBRARY_PATH', '')
os.environ['NCCL_P2P_DISABLE']            = '1'
os.environ['NCCL_IB_DISABLE']             = '1'
os.environ['NCCL_IGNORE_DISABLED_P2P']    = '1'
os.environ['VLLM_WORKER_MULTIPROC_METHOD']= 'spawn'
print("✅ Đã thiết lập biến môi trường.")

# ==============================================================================
# 🚀 BƯỚC 2: KHỞI TẠO VLLM (tối ưu)
# ==============================================================================
from vllm import LLM, SamplingParams

model_name = "Qwen/Qwen2.5-14B-Instruct-AWQ"
print(f"🚀 Đang khởi động vLLM: {model_name}")

llm = LLM(
    model=model_name,

    # ─── Quantization ────────────────────────────────────────────────────────
    # "awq_marlin" dùng Marlin kernel — nhanh hơn "awq" ~30-50% trên GPU Ampere+
    quantization="awq_marlin",
    dtype="bfloat16",           # bfloat16 ổn định hơn float16 cho Qwen2.5

    # ─── Memory / Context ────────────────────────────────────────────────────
    max_model_len=32768,        # Qwen2.5-14B hỗ trợ 128K; 32K đủ cho đề cương
    gpu_memory_utilization=0.92,

    # ─── Speed optimisations ─────────────────────────────────────────────────
    # Prefix caching: toàn bộ file đề cương đều dùng chung 1 system prompt dài.
    # → Phần KV-cache của system prompt chỉ tính 1 lần, tiết kiệm lớn.
    enable_prefix_caching=True,

    # Chunked prefill: chia prefill dài thành chunk, giảm OOM với context lớn.
    enable_chunked_prefill=True,

    # enforce_eager=False: bật CUDA Graph cho decode phase → tăng tốc ~20%
    enforce_eager=False,

    tensor_parallel_size=1,
    trust_remote_code=True,
)
```

# Infer

```python
import json
import os
from docx import Document
from tqdm.notebook import tqdm

# ==========================================
# 1. HÀM ĐỌC DOCX (TEXT + TABLES)
# ==========================================
def read_docx_optimized(file_path):
    """
    Đọc file docx, lấy cả văn bản và nội dung trong bảng.
    Gộp lại thành string sạch để tiết kiệm token.
    """
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"Không tìm thấy file: {file_path}")

    doc = Document(file_path)
    full_text = []

    # 1. Đọc các đoạn văn (Paragraphs)
    for para in doc.paragraphs:
        text = para.text.strip()
        if text:
            full_text.append(text)

    # 2. Đọc các bảng (Tables) - Rất quan trọng với đề cương môn học
    for table in doc.tables:
        for row in table.rows:
            row_text = " | ".join(cell.text.strip() for cell in row.cells)
            if row_text.strip():
                full_text.append(f"| {row_text} |")

    return "\n".join(full_text)
```

# Extract

```python
import json
import os
import time
from vllm import SamplingParams
# ✅ Import đúng class dựa trên kết quả dir() của bạn
from vllm.sampling_params import StructuredOutputsParams 
from tqdm.notebook import tqdm

# ──────────────────────────────────────────────────────────────────────────────
# System prompt — dùng chung cho toàn bộ file
# ──────────────────────────────────────────────────────────────────────────────

SYSTEM_PROMPT = (
    "You are an expert Job Analyst and HR Specialist.\n\n"
    "Your task is to analyze Vietnamese course syllabus documents and extract structured data.\n\n"

    "=== SKILL EXTRACTION RULES ===\n"
    "1. Extract ALL distinct skills — do NOT merge separate skills into one.\n"
    "   BAD: 'Cryptography' covering both symmetric AND asymmetric encryption.\n"
    "   GOOD: Two separate entries: 'Symmetric Encryption' and 'Public Key Cryptography'.\n"
    "2. GENERALIZE academic phrasing into standard professional skill names (1-4 words).\n"
    "   BAD: 'Detect, analyze and evaluate weaknesses of network systems'.\n"
    "   GOOD: 'Network Vulnerability Assessment'.\n"
    "3. PRIORITIZE the CLO table (Chuẩn đầu ra học phần / Course Learning Outcomes) as the\n"
    "   primary source of skills. Each CLO row typically maps to one or more skills.\n"
    "4. Also extract skills from: course description (Mô tả học phần), main content\n"
    "   (Nội dung chính), and weekly lesson plan (Kế hoạch dạy học).\n"

    "=== CATEGORY CLASSIFICATION ===\n"
    "- Hard Skill: domain knowledge, technical or academic competencies.\n"
    "  Examples: Network Security, Cryptography, SQL, Machine Learning.\n"
    "- Soft Skill: interpersonal, cognitive, or behavioral competencies.\n"
    "  Examples: Teamwork, Critical Thinking, Technical Writing, Self-learning.\n"
    "- Tool/Software: a named tool, platform, language, or software product.\n"
    "  Examples: Wireshark, Python, Solidity, Ethereum, MySQL.\n\n"

    "Output in ENGLISH."
)

def _build_messages(context_text: str) -> list:
    return [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": context_text},
    ]

from docx import Document
from docx.table import Table
from docx.text.paragraph import Paragraph
from docx.oxml.ns import qn

def read_docx_optimized(file_path):
    try:
        doc = Document(file_path)
        parts = []

        # Duyệt qua các element con của body theo đúng thứ tự xuất hiện
        for child in doc.element.body:
            if child.tag == qn("w:p"):
                # Paragraph
                para = Paragraph(child, doc)
                text = para.text.strip()
                if text:
                    parts.append(text)
            elif child.tag == qn("w:tbl"):
                # Table
                table = Table(child, doc)
                for row in table.rows:
                    row_text = " | ".join(cell.text.strip() for cell in row.cells)
                    if row_text.strip():
                        parts.append(f"| {row_text} |")

        return "\n".join(parts)
    except Exception as e:
        return ""


# ──────────────────────────────────────────────────────────────────────────────
# BATCH xử lý toàn bộ folder
# ──────────────────────────────────────────────────────────────────────────────
def process_course_folder(input_root_dir: str, output_root_dir: str, llm_instance, complex_schema):
    """
    Xử lý Batch sử dụng tham số 'structured_outputs' cho vLLM version của bạn
    """
    # ── Bước 1: Thu thập file ─────────────────────────────────────────────
    pending = []
    for root, _, files in os.walk(input_root_dir):
        for filename in sorted(files):
            if not filename.endswith(".docx") or filename.startswith("~$"):
                continue
            inp = os.path.join(root, filename)
            rel = os.path.relpath(root, input_root_dir)
            out_dir = os.path.join(output_root_dir, rel)
            os.makedirs(out_dir, exist_ok=True)
            out = os.path.join(out_dir, os.path.splitext(filename)[0] + ".json")
            pending.append((inp, out, filename))

    total = len(pending)
    to_process = [(ip, op, fn) for ip, op, fn in pending if not os.path.exists(op)]
    skipped = total - len(to_process)

    print(f"📂 Tìm thấy {total} file — cần xử lý: {len(to_process)} | bỏ qua: {skipped}")
    if not to_process:
        print("✅ Tất cả file đã được xử lý trước!")
        return

    # ── Bước 2: Đọc docx → build message batch ────────────────────────────
    print("\n📖 Đọc file docx...")
    messages_batch = []
    valid_items = []

    for inp, out, fn in tqdm(to_process, desc="Đọc docx", unit="file"):
        text = read_docx_optimized(inp)
        if text.strip():
            messages_batch.append(_build_messages(text))
            valid_items.append((out, fn))

    if not messages_batch:
        print("⚠️ Không có nội dung để xử lý.")
        return

    print(f"\n🚀 Gửi {len(messages_batch)} request lên GPU...")

    # ── Bước 3: Inference với cấu trúc CHUẨN cho version của bạn ──────────
    
    # ✅ FIX: Dùng structured_outputs=StructuredOutputsParams(json=...)
    sp = SamplingParams(
        temperature=0.1,       
        max_tokens=8192,       
        repetition_penalty=1.05,
        # Syntax chuẩn cho vLLM 0.7.0+ hoặc bản của bạn:
        structured_outputs=StructuredOutputsParams(json=complex_schema)
    )

    t0 = time.perf_counter()
    outputs = llm_instance.chat(messages_batch, sp, use_tqdm=True)
    elapsed = time.perf_counter() - t0

    # ── Bước 4: Parse + lưu ───────────────────────────────────────────────
    success = 0
    failed = 0
    
    for (out_path, fn), result in zip(valid_items, outputs):
        try:
            generated_text = result.outputs[0].text
            data = json.loads(generated_text)
            with open(out_path, "w", encoding="utf-8") as f:
                json.dump(data, f, indent=2, ensure_ascii=False)
            success += 1
        except Exception as e:
            # Ghi file lỗi để debug nếu cần
            with open(out_path + ".log", "w") as f: f.write(result.outputs[0].text)
            failed += 1

    print(f"\n{'='*55}")
    print(f"✅ Hoàn tất: {elapsed:.1f}s | Thành công: {success} | Lỗi: {failed}")
    print(f"{'='*55}")
```

```python
import zipfile
import tempfile
import shutil
import os

# ==============================================================================
# ⚙️ CẤU HÌNH — chỉnh 2 dòng này
# ==============================================================================
ZIP_INPUT_PATH = "Viện Ngân hàng Tài chính_2024-20260307T043713Z-1-001.zip"
OUTPUT_DIR     = "Data_Courses_Output"

# ==============================================================================
# 📦 GIẢI NÉN ZIP
# ==============================================================================
if not os.path.exists(ZIP_INPUT_PATH):
    raise FileNotFoundError(f"Không tìm thấy file zip: {ZIP_INPUT_PATH}")

extract_dir = tempfile.mkdtemp(prefix="course_input_")
print(f"📦 Giải nén: {os.path.basename(ZIP_INPUT_PATH)} → {extract_dir}")

with zipfile.ZipFile(ZIP_INPUT_PATH, "r") as zf:
    zf.extractall(extract_dir)
    print(f"   ✅ {len(zf.namelist())} file được giải nén")

# Nếu zip chỉ chứa 1 folder con duy nhất → dùng folder đó làm root
extracted_items = [
    item for item in os.listdir(extract_dir)
    if not item.startswith("__MACOSX")  # bỏ qua metadata macOS nếu có
]
if len(extracted_items) == 1 and os.path.isdir(os.path.join(extract_dir, extracted_items[0])):
    input_dir = os.path.join(extract_dir, extracted_items[0])
else:
    input_dir = extract_dir

print(f"📁 Input dir: {input_dir}\n")

# ==============================================================================
# 🚀 XỬ LÝ
# ==============================================================================
try:
    process_course_folder(input_dir, OUTPUT_DIR, llm, complex_schema)
finally:
    # Dọn dẹp thư mục tạm dù thành công hay lỗi
    shutil.rmtree(extract_dir, ignore_errors=True)
    print(f"\n🗑️  Đã xóa thư mục tạm: {extract_dir}")
```
